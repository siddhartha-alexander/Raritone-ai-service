```python
!PYTHONPATH=/kaggle/working/triposr_deps:$PYTHONPATH python - <<'PY'
import trimesh
import os

obj_path = "/kaggle/working/TripoSR/output/0/mesh.obj"
glb_path = "/kaggle/working/TripoSR/output/0/model.glb"

mesh = trimesh.load(obj_path, force="mesh")
mesh.export(glb_path)

size_mb = os.path.getsize(glb_path) / (1024 * 1024)

print("✅ GLB CREATED SUCCESSFULLY")
print("GLB:", glb_path)
print("Vertices:", len(mesh.vertices))
print("Polygons:", len(mesh.faces))
print("GLB size:", round(size_mb, 2), "MB")
PY
```
